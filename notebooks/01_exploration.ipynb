# PRAXIS — Exploration Notebook

An interactive environment for inspecting the PRAXIS clinical-scribe pipeline:
1. Load and visualise a PriMock57 GP consultation audio sample
2. Transcribe it with **PRAXIS-AUSCULT** (the contributed Conformer-CTC) and the MedASR / Whisper baselines, side-by-side

Requires PriMock57 audio under `data/`. PRAXIS-AUSCULT additionally needs the `best_primock57.pt` checkpoint (see the README "Reviewer setup" section); the baseline engines auto-download from Hugging Face.

```python
import sys
sys.path.insert(0, '..')

from src.speech_to_text import create_engine, EngineType
from src.speech_to_text.benchmarker import Benchmarker
from src.utils.audio_utils import load_audio
import librosa.display
import matplotlib.pyplot as plt
import numpy as np
```

## 1. Load a sample audio file

```python
# Replace with an actual audio path
audio, sr = load_audio('../data/raw/primock57/test_0000.wav')
print(f'Duration: {len(audio)/sr:.1f}s, Sample rate: {sr}Hz')

fig, ax = plt.subplots(figsize=(12, 3))
librosa.display.waveshow(audio, sr=sr, ax=ax)
ax.set_title('Waveform')
plt.tight_layout()
plt.show()
```

## 2. Transcribe with PRAXIS-AUSCULT and the baselines

```python
# PRAXIS-AUSCULT is the contributed model and needs best_primock57.pt locally.
# MedASR and Whisper are baselines that auto-download from Hugging Face.
auscult = create_engine(EngineType.PRAXIS_CONFORMER)
medasr = create_engine(EngineType.MEDASR)
whisper = create_engine(EngineType.WHISPER)

print('PRAXIS-AUSCULT:', auscult.transcribe(audio, sr))
print('MedASR:        ', medasr.transcribe(audio, sr))
print('Whisper:       ', whisper.transcribe(audio, sr))
```
